### 0:15 - 0:30 | A ideia central

Imagine que você precisa calcular o desconto de um produto em 20 lugares diferentes do programa. Copiar e colar o mesmo código 20 vezes é ruim: se a regra mudar, você corrige 20 vezes.

Uma função resolve isso:

- Você define o código uma vez, com um nome (def).
- Você chama esse nome quando precisar.

Se a regra mudar, você altera em um único lugar.

Pense numa máquina de café: você aperta o botão (chama a função), coloca a dose (parâmetro) e recebe o café (retorno). Não precisa saber como ela funciona por dentro.

### Bloco 5: funções com decisões

```python
def classificar_nota(nota):
    if nota >= 7:
        return "Aprovado"
    elif nota >= 5:
        return "Recuperação"
    else:
        return "Reprovado"

print(classificar_nota(8))
print(classificar_nota(6))
print(classificar_nota(3))
```

Quando o Python encontra um return, a função termina na hora.

### Bloco 6: funções com loops

```python
def calcular_media(lista):
    soma = 0
    for numero in lista:
        soma += numero
    return soma / len(lista)

print(calcular_media([8.5, 6.0, 9.0, 4.5]))
print(f"{calcular_media([10, 6, 10]):.2f}")
```

### Bloco 7: valores padrão

```python
def calcular_preco(valor, desconto=0):
    return valor - (valor * desconto / 100)

print(calcular_preco(200))        # sem desconto
print(calcular_preco(200, 10))    # com 10%
print(calcular_preco(200, desconto=25))
```

Se você não passar o argumento, o Python usa o valor padrão. Esse recurso aparece o tempo todo em bibliotecas de IA (por exemplo, temperature=0.7).

### Bloco 8: escopo (onde as variáveis vivem)

```python
def calcular():
    resultado = 100
    print("Dentro:", resultado)

resultado = 0 # para não ficar com erro eu criei essa vairiável fora
calcular()
print("Fora:", resultado)
```

```python
def maior_de_dois(a, b):
    if a > b:
        return a
    else:
        return b

maior_de_dois(3, 5)
```

```python
def contar_maiores(lista, limite):
    itens_maiores = 0

    for n in lista:
        if n > limite:
            itens_maiores += 1

    return itens_maiores



print(contar_maiores([15, 22, 17, 30, 12, 45], 17))
```

```python
def fatorial(n):
    fatorial = 1
    passos = []  # Lista para guardar os números de cada volta

    # Decrementa do número até 1
    for i in range(n, 0, -1):
        fatorial = fatorial * i
        passos.append(str(i))  # Adiciona o número atual na lista (convertido para texto)

    # Junta todos os elementos da lista usando " x " entre eles
    expressao = " x ".join(passos)

    return f"{n}! | {expressao} = {fatorial}"

fatorial(5)
```

Desafio, mini calculadora: crie calcular(a, b, operacao), em que operacao é "+", "-", "*" ou "/". Devolva o resultado e, se for divisão por zero, devolva a mensagem "Erro: divisão por zero". Se a operação for inválida, devolva "Operação inválida".

```python
def calcular(n1, n2, operacao):
    # 1. Valida se a operação é válida usando 'not in'
    if operacao not in ["+", "-", "*", "/"]:
        return "Operação inválida"

    # 2. Verifica divisão por zero apenas quando a operação for '/' e n2 for 0
    if operacao == "/" and n2 == 0:
        return "Erro: divisão por zero"

    # 3. Executa a operação correspondente
    if operacao == "+":
        return n1 + n2
    elif operacao == "-":
        return n1 - n2
    elif operacao == "*":
        return n1 * n2
    elif operacao == "/":
        return n1 / n2

calcular(10, 2, "/")
#calcular(10, 0, "/")
#calcular(10, 0, "*")
#calcular(10, 2, "%")
```
