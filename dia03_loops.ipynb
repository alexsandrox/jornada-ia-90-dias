### 0:15 - 0:30 | A ideia central

Imagine ter que imprimir os números de 1 a 1000. Sem loops, seriam 1000 linhas de print. Com loop, são 2 linhas.

Existem dois tipos:

for: repete um número conhecido de vezes, ou para cada item de uma coleção. (“Para cada aluno da turma, faça isto.”)
while: repete enquanto uma condição for verdadeira. (“Enquanto a senha estiver errada, pergunte de novo.”)

A indentação de 4 espaços continua valendo: tudo que está recuado abaixo do for ou do while faz parte da repetição.

Bloco 3: acumulando valores

Um loop muito comum: somar enquanto repete.

```python
soma = 0

for s in range(1, 6):
  soma = soma + s
  print(f"S = {s}, soma até agora = {soma}")

print("Soma Total:", soma)
```

Acompanhe a saída linha por linha e veja a variável soma crescendo a cada volta.

Bloco 4: while

```python
contador = 1

while contador <= 5:
  print("Contador = ", contador)
  contador = contador + 1

print("Fim")
```

Atenção: se você esquecer a linha contador = contador + 1, o loop nunca termina (loop infinito). Se acontecer no Colab, clique no botão de parar (quadrado) da célula.

Bloco 5: while com validação (uso real)

```python
# Versão corrigida usando while True (mais recomendada)
while True:
    senha = input("Digite a senha: ")

    if senha == "python123":
        print("Acesso liberado!")
        break

    print("Senha Incorreta!\n")
```

O programa insiste até acertar. É assim que muitos sistemas validam entradas.

Bloco 6: break e continue

```python
# break: interrompe o loop
for i in range(1, 11):
  if i == 6:
    break
  print(i)

# continue: pula somente esta volta
for c in range(1, 11):
  if c % 2 == 0
    continue
  print(c)
```

**Desafio, jogo de adivinhação**: o programa guarda o número secreto 7. Use while para pedir palpites até o usuário acertar, dizendo “maior” ou “menor” a cada erro. No final, mostre em quantas tentativas ele acertou.

**Se travar**: leia o erro, tente por 10 minutos e só então pesquise. Se o loop não parar, é quase sempre porque a variável da condição não está mudando.

```python
# 1. Imprima os números de 1 a 20.

for numero in range(1, 21):
  print(numero)
```

```python
# 2. Imprima apenas os números pares de 1 a 50.

for numero in range(0, 51, 2):
  print(numero)
```

```python
# 3. Peça um número n e mostre a soma de 1 até n.

number = int(input("Digite um número entre 10 e 100: "))
soma = 0

# O range começa em 1 e vai até 'number' (por isso 'number + 1')
for n in range(0, number + 1):
  soma = soma + n # Acumula o valor atual de 'i' na soma

# O print fica fora do loop para mostrar apenas o resultado final
print(f"A soma de 1 até {number} é: {soma}")
```

```python
# 4. Peça um número e calcule o fatorial (5! = 5 × 4 × 3 × 2 × 1 = 120).

numero = int(input("Digite um número: "))
fatorial = 1
passos = []  # Lista para guardar os números de cada volta

# Decrementa do número até 1
for i in range(numero, 0, -1):
    fatorial = fatorial * i
    passos.append(str(i))  # Adiciona o número atual na lista (convertido para texto)
    print("Valor de i =", i)
print(passos)

# Junta todos os elementos da lista usando " x " entre eles
expressao = " x ".join(passos)

print(expressao)

print(f"{numero}! | {expressao} = {fatorial}")
```

```python
# Dada a lista idades = [15, 22, 17, 30, 12, 45], mostre quantas pessoas são maiores de idade e a maior idade da lista (sem usar max()).
idades = [15, 22, 17, 30, 12, 45]
pessoas_maior = 0
maior = idades[0]

for idade in idades:
  if idade >= 18:
    pessoas_maior += 1
  if idade > maior:
    maior = idade

print(f"Maior idade: {maior}")
print(f"Pessoas maior de idade: {pessoas_maior}")
```

```python
# Desafio, jogo de adivinhação
# ----------------------------
# O programa guarda o número secreto 7.
# Use while para pedir palpites até o usuário acertar, dizendo “maior” ou “menor” a cada erro.
# No final, mostre em quantas tentativas ele acertou.

secret_number = 7
tentativas = 0

while True:
  palpite = int(input("Digite o seu palpite: "))
  tentativas += 1

  if palpite == secret_number:
    print(f"Parabéns! 👏 Você acertou em {tentativas} tentativa(s)!")
    break
  elif palpite < secret_number:
    print("O número secreto é MAIOR. Tente novamente!")
  else:
    print("O número secreto é MENOR. Tente novamente!")
```
