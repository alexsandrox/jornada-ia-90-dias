### 0:15 - 0:25 | Planejar o projeto

Antes de escrever código, profissionais planejam. Leia e entenda o que o programa fará:

**Analisador de Texto**: recebe um texto e devolve um relatório com:

- Número de palavras e de caracteres
- As 3 palavras mais frequentes
- A palavra mais longa
- O número de frases

Cada item vira uma função. No final, uma função principal junta todas.

```python
# Planejar o Projeto

## 1. Função que conta palavras
def numero_de_palavras(texto):
    """
    Função com responsabilidade única:
    Recebe um texto e retorna exatamente a quantidade de palavras nele contidas.
    """

    # Passo 1: O método .split() pega a string longa e a divide em pedaços
    # utilizando os espaços em branco como separadores por padrão.
    # Isso transforma a frase em uma lista contendo cada palavra isolada.
    palavras = texto.split()

    # Passo 2: A função nativa len() conta quantos itens existem
    # dentro da lista gerada pelo .split().
    quantidade = len(palavras)

    # Passo 3: O comando return devolve esse número calculado
    # para quem chamou a função (no nosso caso, a 'funcao_principal').
    return quantidade
```

```python
# 2. Função que conta o número de caracteres
def numero_de_caracteres(texto):
    """
    Função com responsabilidade única:
    Recebe um texto e retorna a quantidade total de caracteres (tamanho da string).
    """

    # Passo 1: A função nativa len() conta quantos elementos (caracteres)
    # existem dentro da string de texto. Em Python, uma string funciona
    # como uma sequência de caracteres.
    total_caracteres = len(texto.replace(" ", ""))

    # Passo 2: O return devolve esse número total para a função principal.
    return total_caracteres
```

```python
## 3. Função que conta as palavras mais frequentes
def tres_palavras_mais_frequentes(texto):
    """
    Função com responsabilidade única:
    Recebe um texto, calcula a frequência de cada palavra e retorna as 3 mais comuns.
    """

    # Passo 1: Padronizamos tudo para minúsculo e dividimos a frase em uma lista de palavras.
    # Usar o .lower() evita que "O" e "o" sejam contados como palavras diferentes.
    palavras = texto.lower().split()

    # Passo 2: Criamos um dicionário vazio para armazenar a contagem de cada palavra.
    frequencias = {}

    # Passo 3: Percorremos a lista de palavras contando-as com a ajuda do .get()
    for palavra in palavras:
        # Ignora palavras com 2 letras ou menos (ex: "o", "a", "de")
        if len(palavra) > 2:
            frequencias[palavra] = frequencias.get(palavra, 0) + 1

    # Passo 4: Ordenamos o dicionário da palavra mais frequente para a menos frequente.
    # O comando .items() transforma o dicionário em uma lista de pares (palavra, quantidade).
    # O 'key=lambda x: x[1]' avisa o Python: "ordene com base no segundo item de cada par" (ou seja, a contagem).
    # O 'reverse=True' garante que a ordem seja decrescente (do maior para o menor).
    palavras_ordenadas = sorted(frequencias.items(), key=lambda x: x[1], reverse=True)

    # Passo 5: Usamos o slicing [:3] para pegar apenas os 3 primeiros elementos da lista ordenada.
    top_3 = palavras_ordenadas[:3]


    # Passo 6: Devolvemos o resultado para a função principal.
    return top_3

texto_teste = "A Inteligência Artificial está mudando o mundo. Quem aprende IA hoje ganha vantagem. Aprender com prática é o melhor caminho."


tres_palavras_mais_frequentes(texto_teste)
```

```python
## 4. Função que separa a palavra mais longa
def palavra_mais_longa(texto):
    """
    Função com responsabilidade única:
    Recebe um texto, divide em palavras e retorna a que possui o maior número de caracteres.
    """
    # Passo 1: Divide o texto em uma lista de palavras usando os espaços como separador
    palavras = texto.split()

    # Passo 2: Se o texto estiver vazio, evitamos um erro retornando uma string vazia
    if not palavras:
        return ""

    # Passo 3: Usamos a função max() para encontrar o maior elemento da lista.
    # O argumento 'key=len' diz ao Python: "Compare as palavras com base no tamanho delas (len)".
    maior_palavra = max(palavras, key=len)

    # Passo 4: Retorna a palavra mais longa encontrada
    return maior_palavra
```

```python
# Função que conta o número de frases
def contar_frases(texto):
    # Conta quantas vezes o ponto final aparece no texto
    qtd_pontos = texto.count('.')

    # Conta quantas vezes o ponto de exclamação aparece no texto
    qtd_exclamacoes = texto.count('!')

    # Conta quantas vezes o ponto de interrogação aparece no texto
    qtd_interrogacoes = texto.count('?')

    # Conta quantas vezes o ponto e vírgula aparecem no texto
    qtd_ponto_virgula = texto.count(';')

    # Conta quantas vezes os dois pontos aparece no texto
    qtd_dois_pontos = texto.count(':')

    # Soma todas as pontuações finais para encontrar o total de frases
    total_frases = qtd_pontos + qtd_exclamacoes + qtd_interrogacoes + qtd_ponto_virgula + qtd_dois_pontos

    # Retorna o resultado da contagem
    return total_frases


# --- Testando com o texto fornecido ---
texto = "A inteligência artificial está mudando o mundo. Quem aprende IA hoje ganha vantagem. Aprender com prática é o melhor caminho."

# Chamando a função e exibindo o resultado
resultado = contar_frases(texto)
print(f"Número de frases: {resultado}")
```

```python
# Função que chama todas as outras e mostra o relatório
def analisar(texto):
    # Remove espaços extras nas pontas do texto recebido
    texto_limpo = texto.strip()

    # [BOA PRÁTICA] Verifica se o texto está vazio após a limpeza para evitar quebrar o programa
    if not texto_limpo:
        # Retorna um dicionário indicando que o texto está vazio/zerado
        return {
            "Aviso": "O texto fornecido está vazio.",
            "Contagem de palavras": 0,
            "Contagem de caracteres": 0,
            "Palavras mais frequentes": [],
            "Palavra mais longa": "",
            "Número de frases": 0
        }

    # Chama as funções auxiliares para obter as métricas
    palavras = numero_de_palavras(texto_limpo)
    caracteres = numero_de_caracteres(texto_limpo)
    frequentes = tres_palavras_mais_frequentes(texto_limpo)
    longa = palavra_mais_longa(texto_limpo)
    frases = contar_frases(texto_limpo)

    # Cria o dicionário onde as chaves já são descritivas para a exibição
    relatorio = {
        "Contagem de palavras": palavras,
        "Contagem de caracteres": caracteres,
        "Palavras mais frequentes": frequentes,
        "Palavra mais longa": longa,
        "Número de frases": frases
    }

    # Retorna o dicionário completo
    return relatorio


# --- Bloco de Testes e Execução ---

# Teste 1: Testando com o texto padrão exigido pelo plano
texto_exemplo = "A inteligência artificial está mudando o mundo. Quem aprende IA hoje ganha vantagem. Aprender com prática é o melhor caminho."

print("--- TESTE 1: Texto Padrão ---")
resultado_padrao = analisar(texto_exemplo)
for chave, valor in resultado_padrao.items():
    print(f"{chave.ljust(26)}: {valor}")

print("\n" + "="*40 + "\n")

# Teste 2: Testando com texto vazio ("") para conferir o comportamento do 'if not texto_limpo'
print("--- TESTE 2: Texto Vazio ---")
resultado_vazio = analisar("")
for chave, valor in resultado_vazio.items():
    print(f"{chave.ljust(26)}: {valor}")

print("\n" + "="*40 + "\n")

# Teste 3: Interação real com o usuário via input()
print("--- TESTE 3: Entrada Dinâmica do Usuário ---")
texto_usuario = input("Digite ou cole um texto para análise: ")
resultado_usuario = analisar(texto_usuario)

print("\nRELATÓRIO DE ANÁLISE DE TEXTO (USUÁRIO)")
print("---------------------------------------")
for chave, valor in resultado_usuario.items():
    print(f"{chave.ljust(26)}: {valor}")
```
