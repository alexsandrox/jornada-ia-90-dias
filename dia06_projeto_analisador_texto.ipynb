### 0:15 - 0:25 | Planejar o projeto

Antes de escrever código, profissionais planejam. Leia e entenda o que o programa fará:

**Analisador de Texto**: recebe um texto e devolve um relatório com:

- Número de palavras e de caracteres
- As 3 palavras mais frequentes
- A palavra mais longa
- O número de frases

Cada item vira uma função. No final, uma função principal junta todas.

```python
# Planejar o Projeto

## 1. Função que limpa palavras
def limpar_palavras(texto):
    """
    Recebe um texto, converte para minúsculas, remove os símbolos
    de pontuação especificados usando replace e divide o texto em uma lista de palavras.
    """
    texto_limpo = texto.lower()
    simbolos = ['.', ',', '!', '?', ';', ':']

    for simbolo in simbolos:
        texto_limpo = texto_limpo.replace(simbolo, '')

    return texto_limpo.split()


def calcular_frequencias(palavras, tamanho_minimo=3):
    """
    Recebe uma lista de palavras limpas e retorna um dicionário
    com a contagem de frequência, ignorando palavras menores que o tamanho_minimo.
    """
    frequencias = {}
    for palavra in palavras:
        # Filtra palavras cujo comprimento seja menor que o limite mínimo
        if len(palavra) >= tamanho_minimo:
            frequencias[palavra] = frequencias.get(palavra, 0) + 1
    return frequencias


def top_palavras(frequencias, n=3):
    """
    Recebe um dicionário de frequências e o número 'n',
    retornando as 'n' palavras mais frequentes ordenadas de forma decrescente.
    """
    palavras_ordenadas = sorted(frequencias.items(), key=lambda x: x[1], reverse=True)
    return palavras_ordenadas[:n]


def palavra_mais_longa(palavras):
    """
    Recebe uma lista de palavras limpas e retorna a palavra mais longa.
    """
    if not palavras:
        return ""

    mais_longa = palavras[0]
    for palavra in palavras:
        if len(palavra) > len(mais_longa):
            mais_longa = palavra

    return mais_longa


def numero_de_palavras(palavras):
    """
    Recebe a lista de palavras limpas e retorna a quantidade total.
    """
    return len(palavras)


def numero_de_caracteres(texto):
    """
    Recebe o texto original e retorna o tamanho total da string (incluindo espaços).
    """
    return len(texto)


def contar_frases(texto):
    """
    Recebe o texto original e conta o número de frases apenas através de ., ! e ?.
    """
    total = 0
    for simbolo in [".", "!", "?"]:
        total += texto.count(simbolo)
    return total


def analisar(texto):
    """
    Função principal que coordena todas as análises.
    """
    if not texto or not texto.strip():
        return {
            "Aviso": "O texto fornecido está vazio.",
            "Contagem de palavras": 0,
            "Contagem de caracteres": 0,
            "Palavras mais frequentes": [],
            "Palavra mais longa": "",
            "Número de frases": 0
        }

    palavras_limpas = limpar_palavras(texto)
    freq = calcular_frequencias(palavras_limpas)

    relatorio = {
        "Contagem de palavras": numero_de_palavras(palavras_limpas),
        "Contagem de caracteres": numero_de_caracteres(texto),
        "Palavras mais frequentes": top_palavras(freq, n=3),
        "Palavra mais longa": palavra_mais_longa(palavras_limpas),
        "Número de frases": contar_frases(texto)
    }

    return relatorio
```

```python
# Teste 1: Contagem de frases com ponto e vírgula / dois-pontos
print("--- TESTE 1: contar_frases ---")
print(contar_frases("Ingredientes: ovos; leite. Misture."))  # Deve retornar 2

print("\n" + "="*40 + "\n")

# Teste 2: Frequência com filtro de tamanho mínimo
print("--- TESTE 2: top_palavras com filtro ---")
print(top_palavras(calcular_frequencias(limpar_palavras("Aprender IA é bom. Aprender é bom. Aprender sempre."))))  # Deve retornar [('aprender', 3), ('bom', 2), ('sempre', 1)]

print("\n" + "="*40 + "\n")

# Teste 3: Relatório completo do texto padrão
texto = "A inteligência artificial está mudando o mundo. Quem aprende IA hoje ganha vantagem. Aprender com prática é o melhor caminho."
print("--- TESTE 3: analisar(texto) ---")
resultado_padrao = analisar(texto)
for chave, valor in resultado_padrao.items():
    print(f"{chave.ljust(26)}: {valor}")

print("\n" + "="*40 + "\n")

# Teste 4: Texto vazio
print("--- TESTE 4: analisar('') ---")
resultado_vazio = analisar("")
for chave, valor in resultado_vazio.items():
    print(f"{chave.ljust(26)}: {valor}")
```
