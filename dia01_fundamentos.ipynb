```python
# Peça o nome e responda: “Olá, [nome]! Bem-vindo à jornada de IA.”
nome = input("Qual o seu nome? ")
print(f"Olá, {nome}! Bem-vindo à jornada de IA.")
```

```python
# Peça dois números e mostre soma, subtração, multiplicação e divisão.
number1 = int(input("Digite o primeiro número: "))
number2 = int(input("Digite o segundo número: "))
print(f"A soma dos números é: {number1 + number2}")
print(f"A subtração dos números é: {number1 - number2}")
print(f"A multiplicação dos números é: {number1 * number2}")
print(f"A divisão dos números é: {number1 / number2}")
```

```python
# Peça a idade e calcule o ano de nascimento aproximado (ano atual: 2026).
idade = int(input("Qual sua idade? "))
anoNascimento = 2026 - idade

print(f"Você tem {idade} anos e nasceu em: {anoNascimento}")
```

```python
# Peça um valor em reais e converta para dólar, usando uma cotação fixa que você escolher.
valor = float(input("DIgite um valor em reais R$ "))
dolar = 5.51
convertido = valor / dolar

print(f"O valor de R$ {valor} convertido para dolar hoje é US$: {convertido}")
```

```python
#Desafio
price = float(input("Digite o preço do produto: "))
discount = float(input("Digite a porcentagem de desconto: "))
finalPrice = price - (price * (discount / 100))
print(f"O preço final do produto com desconto de {discount}% é: R$", finalPrice)
```
