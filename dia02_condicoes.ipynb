```python
# 0:30 - 1:30 | Guia prático (digite tudo)
# Bloco 1: operadores de comparação
# Cada comparação devolve True ou False.
idade = 47
print(f"{idade} é maior que 25: {idade > 25}")  # maior que
print(f"{idade} é menor que 25: {idade < 25}")  # menor que
print(f"{idade} é maior ou igual a 25: {idade >= 25}")  # maior ou igual
print(f"{idade} é menor ou igual a 25: {idade <= 25}")  # menor ou igual
print(f"{idade} é igual a 25: {idade == 25}")  # igual (atenção: dois sinais!)
print(f"{idade} é diferente de 25: {idade != 25}")  # diferente

# Erro clássico:
# = guarda um valor (recebe)
# == compara dois valores.
```

```python
# Bloco 2: if e else
idade = int(input("Qual sua idade? R: "))

if idade >= 18:
  print("Você é maior de idade!")
else:
  print("Você é menor de idade!")
```

```python
# Bloco 3: elif (várias opções)
nota = float(input("Qual sua nota? R: "))

if nota >= 7:
  print("Aprovado!")
elif nota >= 5:
  print("Recuperação")
else:
  print("Reprovado!")
```

```python
# Bloco 4: and, or, not
idade = 47
tem_documento = False

if idade >= 18 and tem_documento:
  print("Entrada Liberada!")

if idade < 12 or idade > 65:
  print("Tem direito a desconto")

if not tem_documento:
  print("Precisa apresentar documento")

# and: as duas condições precisam ser verdadeiras
# or: basta uma
# not: inverte
```

```python
# Bloco 5: comparando textos
senha = input("Digite a senha : ")

if senha.lower() == "python123":
  print("Acesso liberado!")
else:
  print("Senha incorreta!")
```

```python
# Peça um número e diga se ele é par ou ímpar (dica: use %).
numero = int(input("Digite um número : "))

if (numero % 2) == 0:
  print("O número é par!")
else:
  print("O número é impar!")
```

```python
# Peça dois números e mostre qual é o maior (ou se são iguais).
numero_1 = int(input("Digite o primeiro número: "))
numero_2 = int(input("Digite o segundo número: "))

if numero_1 > numero_2:
  print(f"O número {numero_1} é maior que o numero {numero_2}")
elif numero_1 < numero_2:
  print(f"O número {numero_1} é menor que o número {numero_2}")
else:
  print(f"Os números {numero_1} e {numero_2} são iguais")
```

```python
# Peça a nota e classifique: A (9 ou mais), B (7 a 8.9), C (5 a 6.9), D (abaixo de 5).
nota = float(input("Digite a nota: "))

if nota >= 9:
  print("A")
elif nota >= 7:
  print("B")
elif nota >= 5:
  print("C")
else:
  print("D")
```
