### 0:15 - 0:30 | A ideia central

Numa lista, você acessa os itens pela posição (frutas[0]). Mas e se quiser guardar os dados de uma pessoa? Nome, idade, cidade... Lembrar que a posição 2 é a cidade é ruim.

O dicionário resolve isso: cada valor tem um nome (chamado chave). Funciona como uma ficha cadastral:

Chave = o rótulo ("nome")
Valor = o conteúdo ("Ana")

Já a tupla é como uma lista que não pode ser alterada depois de criada, boa para dados fixos, como coordenadas.

E o texto (str) tem dezenas de ferramentas prontas para limpar, separar e transformar. Em IA, você vai passar boa parte do tempo preparando texto.

• Lista [....]

• Tupla (....)

• Conjunto (set) {....}

•  Dicionário (dict) {chave: valor}

### Bloco 3: get() e verificação de chave

```python
pessoa = {"nome": "Ana", "idade": 30}

print(pessoa.get("nome"))
print(pessoa.get("profissao"))            # None, sem erro
print(pessoa.get("profissao", "Não informado"))

if "idade" in pessoa:
    print("Tem idade cadastrada")
```

O get() é a forma segura de ler: se a chave não existir, devolve um valor padrão em vez de dar erro.

### Bloco 4: percorrendo um dicionário

```python
notas = {"Ana": 8.5, "Carlos": 6.0, "Bia": 9.0}

for nome in notas:
    print(nome)

# O .items() entrega chave e valor juntos, e é o mais usado.
for nome, nota in notas.items():    # buscando duas informações
    print(f"{nome} tirou {nota}")

print(notas.keys())     # Pegar as chaves
print(notas.values())   # Pegar os valores
```

O .items() entrega chave e valor juntos, e é o mais usado.

### Bloco 5: lista de dicionários

```python
alunos = [
    {"nome": "Ana", "nota": 8.5},
    {"nome": "Carlos", "nota": 6.0},
    {"nome": "Bia", "nota": 9.0}
]

# Repare nas aspas simples dentro da f-string, que evitam conflito com as aspas duplas externas.
for aluno in alunos:
    print(f"{aluno['nome']}: {aluno['nota']}") # usando aspas simples

print(alunos[0]["nome"])
```

Esta é a estrutura mais comum em IA: uma lista de registros. As respostas de APIs, os datasets e as conversas com modelos têm esse formato. Repare nas aspas simples dentro da f-string, que evitam conflito com as aspas duplas externas.

### Bloco 8: métodos de texto

```python
texto = "  Olá, Mundo da IA!  "

print(texto.lower())
print(texto.upper())
print(texto.strip())    # remove espaços das pontas esquerdas e direitas
print(texto.replace("IA", "Inteligência Artificial"))
print(len(texto))
```

Agora separando e juntando:

```python
frase = "python é muito útil para ia"

# Separando a frase e criando uma lista
palavras = frase.split()

# Imprimindo a frase
print(palavras)
print("---------")

# contando quantas palavras a frase contém
print(len(palavras))
print("---------")

# juntando as palavras
print("-".join(palavras))
print("---------")

# Apenas a primeira letra da frase em maiúsculo
print(frase.capitalize())
print("---------")

# Verificando se a palavra está na frase
print("python" in frase)
print("---------")

# Todas as iniciais em maiúsculo
print(frase.title())
print("---------")

# contando quantas letras 'a' a frase contém
print(frase.count("a"))
```

- split() quebra o texto em uma lista de palavras.
- join() faz o contrário.

### Bloco 9: fatiando texto (slicing)

```python
palavra = "inteligencia"

print(f"A Palavra é.............: {palavra}")
print("--------------------------------------")
print(f"1. Primeira Letra.......: {palavra[0]}")
print(f"2. Última Letra.........: {palavra[-1]}")
print(f"3. 5 primeiras..........: {palavra[0:5]}")     # "intel" (o fim não entra, como no range)
print(f"4. Até a 5 letra........: {palavra[:5]}")      # até 0 5º caracter
print(f"5. Depois da 5º.........: {palavra[5:]}")      # depois do 5º caracter
print(f"6. De trás para frente..: {palavra[::-1]}")    # invertido (de trás para frente)
```

Texto funciona como uma lista de letras, e a regra do fim não incluído é a mesma do range.

### Bloco 10: JSON na prática

```python
import json

texto_json = '{"nome": "Ana", "idade": 30, "habilidades": ["Python", "SQL"]}'

# texto -> dicionário
# loads() converte texto para dicionário
dados = json.loads(texto_json)

print(dados["nome"])
print(dados["habilidades"][0])

# alteração de dados
dados["idade"] = 31

# dicionário -> texto
# dumps() converte o dicionário em texto (inverso do loads())
# o formato do texto é JSON
# Definindo ensure_ascii=False, você força o Python a manter os caracteres
# originais legíveis na string resultante, preservando acentos e pontuações do nosso idioma.
novo_json = json.dumps(dados, ensure_ascii=False)

print(novo_json)
```

APIs de IA enviam e recebem dados nesse formato. loads converte texto em dicionário, e dumps faz o caminho inverso.

```python
# 3. Crie uma função contar_palavras(texto) que devolve quantas palavras tem um texto.
def contar_palavras(texto):

    # para contar palavras, primeiro precisamos transformar o texto em uma lista
    # vamos separar as palavras com o método split()
    palavras = texto.split()

    print(f"Quantidade de Palavras: {len(palavras)}")

contar_palavras("Estou aprendendo python para ajudar empresas")
```

```python
# 4. Crie uma função eh_palindromo(palavra) que devolve True se a palavra é igual ao contrário (ex.: "arara"). Use slicing.
def eh_palindromo(palavra):

    # Normalizando a palavra ou frase para funcionar
    # Remove os espaços e transforma tudo em minúsculas
    palavra = palavra.replace(" ", "").lower()

    # usando slicing [::-1] (De trás para frente)
    return palavra == palavra[::-1]

# verificar se a palavra é um palíndromo
if eh_palindromo("Luz Azul"):
    print("SIM é um Palíndromo")
else:
    print("NÃO é um Palíndromo")
```

6. **Desafio**, contador de palavras: crie frequencia(texto) que devolve um dicionário com quantas vezes cada palavra aparece.
Teste com "o gato viu o rato e o rato fugiu". Resultado esperado: {'o': 3, 'gato': 1, 'viu': 1, 'rato': 2, 'e': 1, 'fugiu': 1}. Dica: use split(), um for e o get() com valor padrão 0.

```python
# 6. Desafio
def frequencia(texto):
    # 1. Cria um dicionário vazio que vai armazenar a contagem de cada palavra
    contagem_palavras = {}

    # 2. Divide a string 'texto' em uma lista de palavras, usando o espaço como separador
    palavras = texto.lower().split()

    # 3. Percorre cada palavra individualmente dentro da lista gerada
    for palavra in palavras:
        # 4. O método .get(palavra, 0) procura a palavra no dicionário:
        #    - Se a palavra JÁ EXISTE, ele retorna o valor atual dela.
        #    - Se a palavra NÃO EXISTE AINDA, ele retorna 0 (evitando um KeyError).
        #    Em seguida, somamos 1 e salvamos de volta no dicionário.
        contagem_palavras[palavra] = contagem_palavras.get(palavra, 0) + 1

    # 5. Retorna o dicionário pronto com todas as frequências contadas
    return contagem_palavras

# --- Testando a função com o exemplo do desafio ---
texto_teste = "o gato viu o rato e o rato fugiu"
resultado = frequencia(texto_teste)

print(resultado)
```
