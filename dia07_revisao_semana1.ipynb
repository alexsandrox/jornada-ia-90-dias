### Parte B: escrever código
- Escreva contar_vogais(texto), que devolve quantas vogais (a, e, i, o, u) o texto tem. Use lower() e um for.

- Escreva somente_pares(lista), que devolve uma nova lista só com os números pares.

- Dada a lista alunos = [{"nome": "Ana", "nota": 8}, {"nome": "Leo", "nota": 5}, {"nome": "Bia", "nota": 9}], escreva um código que imprima só o nome de quem tem nota maior ou igual a 7.

- Escreva inverter_palavras(frase), que devolve a frase com a ordem das palavras invertida ("eu amo python" vira "python amo eu"). Dica: split(), slicing e join().

```python
# Função para contar vogais
def contar_vogais(texto):
    vogais = "aeiouáàâãéêíóôõú"
    contador = 0

    for letra in texto.lower():
        if letra in vogais:
            contador += 1

    return contador

print(f"A quantidade de vogais é: {contar_vogais('Estou aprendendo Python, e isso é o máximo!')}")
```

```python
# Função que devolve uma lista com apenas os números pares
def somente_pares(lista):
    pares = []

    for numero in lista:
        if numero % 2 == 0:
            # usando append para preencher a lista
            pares.append(numero)

    return pares

print(f"A lista de números pares é: {somente_pares([1, 2, 3, 4, 5, 6, 7, 9, 11, 12, 15])}")
```

```python
# 1. Função que inverte as palavras de uma frase
def inverter_palavras(frase):

    # frase.split()	["Eu", "amo", "python"] (lista)
    palavras = frase.split()

    # palavras[::-1]	["python", "amo", "Eu"] (lista invertida)
    palavras_invertidas = palavras[::-1]

    # " ".join(...)	"python amo Eu" (texto)
    frase_invertida = " ".join(palavras_invertidas)

    return frase_invertida

print(f"A frase invertida é: {inverter_palavras('Eu amo python')}")
```

```python
# 2. VERSÃO ENXUTA : Função que inverte as palavras de uma frase
def inverter_palavras(frase):
    return " ".join(frase.split()[::-1])

print(f"A frase invertida é: {inverter_palavras('Eu amo python')}")
```

```python
def media(lista):
    # Verifica se a lista está vazia para evitar erro de divisão por zero
    if not lista:
        return 0

    soma = 0
    for n in lista:
        soma += n

    # O 'return' agora está fora do loop, executado apenas após a soma completa
    return soma / len(lista)

print(media([4, 6, 8]))
```

**Desafio**: escreva eh_anagrama(a, b), que devolve True se duas palavras têm as mesmas letras em ordem diferente ("amor" e "roma"). Dica: sorted() funciona em textos.

```python
# Função que verifica se as palavras são um anagrama
def eh_anagrama(a, b):
    return sorted(a.lower()) == sorted(b.lower())

if eh_anagrama("amor", "roma"):
    print("São anagramas")
else:
    print("Não são anagramas")

print(eh_anagrama("amor", "roman"))   # False
```
