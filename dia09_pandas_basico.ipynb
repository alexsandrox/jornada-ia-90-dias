Bloco 1: criando um DataFrame

```python
import pandas as pd

dados = {
    "nome": ["Ana", "Leo", "Bia", "Alex", "Leon"],
    "idade": [28, 35, 22, 47, 50],
    "cidade": ["São Paulo", "Belo Horizonte", "Manaus", "Mogi das Cruzes", "São Paulo"],
    "salario": [4500, 6200, 3800, 7100, 5500]
}

df = pd.DataFrame(dados)
print(df)
print()

# No Colab, escrever só df na última linha da célula mostra uma tabela formatada.
df
```

Bloco 2: conhecendo a tabela

```python
# Estes comandos são o primeiro passo ao abrir qualquer dataset novo.
# Funcionam como um "check-up médico" inicial antes de limpar, filtrar ou treinar qualquer modelo de IA.

# 1. Dimensões da tabela (.shape):
# Retorna uma tupla no formato (quantidade_de_linhas, quantidade_de_colunas).
# É um atributo (não leva parênteses) herdado diretamente do NumPy.
print("Linhas e Colunas :", df.shape)       # (linhas, colunas)

# 2. Nomes das colunas (.columns):
# Exibe uma lista/índice com o nome exato de cada coluna da tabela.
# Muito útil para detectar espaços escondidos nos nomes (ex: " salario" em vez de "salario") antes de filtrar.
print("Colunas .........:", df.columns)

print("\n")
print("Tipos de dados")
print("="*14)
# 3. Tipagem das colunas (.dtypes):
# Mostra como o pandas interpretou cada coluna:
# - int64 / float64: números inteiros e decimais (prontos para cálculos matemáticos).
# - object / string: textos (ou colunas numéricas que vieram "sujas" com algum texto misturado).
# - bool: valores lógicos True/False.
print(df.dtypes)

print("\n")
print("Primeiras 2 linhas")
print("="*18)
# 4. Amostra visual do topo (.head(n)):
# Exibe as primeiras 'n' linhas da tabela (se não passar número, o padrão é 5).
# Serve para conferir visualmente se o cabeçalho e os dados alinharam corretamente na leitura.
print(df.head(2))    # primeiras 2 linhas

print("\n")
print("Informações do DataFrame")
print("="*24)
# 5. Raio-X estrutural completo (.info()):
# Mostra de uma só vez: o total de linhas, os nomes das colunas, quantos valores NÃO-NULOS existem
# em cada coluna (essencial para bater o olho e achar dados faltantes), os tipos de dados e o uso de memória RAM.
# Detalhe importante: o .info() já imprime o resultado direto na tela por conta própria, por isso NÃO vai dentro de print().
df.info()

print("\n")
print("Estatísticas das colunas numéricas")
print("="*34)
# 6. Resumo estatístico automático (.describe()):
# Ignora colunas de texto e calcula 8 métricas para todas as colunas numéricas de uma vez:
# - count: contagem de valores válidos (não nulos).
# - mean / std: média e desvio padrão (dispersão).
# - min / max: menor e maior valor (ótimo para flagrar erros grotescos, como idade = -5 ou 999).
# - 25%, 50% (mediana), 75%: os quartis que mostram como os dados estão distribuídos.
print(df.describe())  # estatísticas das colunas numéricas
```

### Detalhe técnico importante nesse bloco:

Se você escrever print(df.info()), o Python vai exibir todo o relatório estrutural normalmente e, logo abaixo dele, imprimir a palavra None — porque a função .info() já escreve direto na saída padrão e não retorna nenhum valor.

Bloco 3: selecionando colunas

```python
# 1. Seleção de uma única coluna (retorna uma Series):
# Ao passar o nome de uma coluna entre colchetes simples df["nome"], o pandas extrai
# uma estrutura unidimensional chamada Series (como uma coluna isolada do Excel, com índice à esquerda e valores à direita).
# Como é uma Series, você pode aplicar direto nela métodos como .mean(), .max() ou .value_counts().
print("Selecionando uma coluna")
print("="*23)
print(df["nome"])              # uma coluna (Series)

# 2. Seleção de múltiplas colunas (retorna um novo DataFrame):
# Observe os colchetes duplos df[["nome", "salario"]]:
# - Os colchetes de FORA são o operador de seleção do DataFrame (df[...]).
# - Os colchetes de DENTRO criam uma lista Python comum contendo os nomes das colunas desejadas: ["nome", "salario"].
# Como você pediu uma lista de colunas (mesmo que a lista tenha apenas 1 item, ex: df[["nome"]]),
# o pandas mantém o formato bidimensional de tabela (DataFrame) em vez de reduzir para uma Series.
print("\n")
print("Selecionando várias colunas")
print("="*27)
print(df[["nome", "salario"]]) # várias colunas (note os colchetes duplos)
```

### Diferença prática entre df["nome"] e df[["nome"]]:
- df["nome"] (1 par de colchetes): Devolve uma Series (1 dimensão — vetor). Ideal para fazer contas (df["salario"].mean()) ou criar máscaras de filtro (df["idade"] > 30).

- df[["nome"]] (2 pares de colchetes): Devolve um DataFrame (2 dimensões — tabela). Ideal quando você quer recortar um subconjunto de colunas para exibir como tabela ou passar como entrada para algoritmos de Machine Learning (que geralmente exigem matrizes 2D).

Bloco 4: estatísticas por coluna

```python
# Tudo que você fez com NumPy no Dia 8 funciona aqui.
# Como cada coluna de um DataFrame do pandas (chamada de Series) é construída sobre um array NumPy,
# os mesmos métodos matemáticos e estatísticos vetorizados funcionam diretamente nela.

# 1. Seleciona a coluna "salario" usando colchetes (df["salario"]) e calcula a média aritmética (.mean()).
# Por padrão, o pandas ignora automaticamente células vazias (NaN) ao fazer esse cálculo.
print("Cálculo da Média Aritimética")
print("="*28)
print(df["salario"].mean())

# 2. Percorre todos os valores da coluna "salario" de forma vetorizada e retorna o maior valor encontrado (.max()).
print()
print("Mostrar o maior salário encontrado")
print("="*34)
print(df["salario"].max())

# 3. Seleciona a coluna "idade" e retorna o menor valor registrado nessa coluna (.min()).
print()
print("Mostrar o menor valor (idade)")
print("="*29)
print(df["idade"].min())

# 4. Muito usado para colunas de texto/categorias: conta quantas vezes cada valor único aparece na coluna "cidade".
# Já devolve o resultado ordenado da cidade com mais ocorrências para a com menos ocorrências.
print()
print("Mostra e ordena a cidade com mais ocorrências")
print("="*45)
print(df["cidade"].value_counts())
```

Detalhe importante para a Engenharia de IA:
- Colunas numéricas (mean, max, min): Funcionam exatamente como no NumPy, com a vantagem de que o pandas já ignora valores nulos (NaN) automaticamente sem travar o cálculo (no NumPy puro, um único NaN faria .mean() retornar nan).

- Colunas categóricas/texto (value_counts): É um dos comandos mais usados na análise exploratória de dados (EDA) para descobrir rapidamente se os seus dados estão desbalanceados (por exemplo, se 90% dos registros vêm de uma única cidade ou categoria).

Bloco 5: filtros

```python
# Bloco 5: filtros
# Assim como fizemos com as máscaras booleanas no NumPy, aqui filtramos linhas inteiras
# do DataFrame passando condições lógicas dentro dos colchetes df[...].

print("Mostrar pessoas com idade maior que 30 anos")
print("===========================================")
# 1. A expressão interna df["idade"] > 30 verifica cada linha e gera uma coluna invisível de True e False.
# 2. Os colchetes externos df[...] mantêm apenas as linhas inteiras onde o resultado foi True.
print(df[df["idade"] > 30])

print()
print("Mostrar pessoas que moram na cidade de SP")
print("=========================================")
# Usa o operador de igualdade (==) para comparar textos.
# Atenção: a comparação é sensível a maiúsculas, minúsculas e acentos ("São Paulo" != "sao paulo").
# Retorna apenas as linhas completas em que a coluna "cidade" contém exatamente "São Paulo".
print(df[df["cidade"] == "São Paulo"])

print()
print("Mostrar pessoas com idade maior que 25 e salário maior que 4k")
print("=============================================================")
# Filtro composto com duas condições simultâneas:
# 1. No pandas (e no NumPy), usamos '&' no lugar de 'and' (e '|' no lugar de 'or') para comparar colunas inteiras.
# 2. Os parênteses em volta de cada condição — (condição 1) & (condição 2) — são OBRIGATÓRIOS,
#    pois o operador '&' tem prioridade maior que '>' na ordem de execução do Python. Sem eles, ocorre um erro (TypeError).
print(df[(df["idade"] > 25) & (df["salario"] > 4000)])
```

É o mesmo filtro por condição do Dia 8, com & e parênteses.

```python
# Importa o módulo nativo 'unicodedata' do Python, que dá acesso ao banco de dados oficial
# de caracteres Unicode (permite identificar letras, acentos, símbolos e categorias gráficas).
import unicodedata

def normalizar_texto(texto):
    """Remove acentos, espaços extras e converte para minúsculo."""

    # Programação defensiva (guard clause): verifica se o valor recebido é realmente uma string (texto).
    # Se a célula da tabela estiver vazia (NaN ou None) ou contiver um número, evita que o código quebre
    # com um erro (TypeError) e devolve uma string vazia "" com segurança.
    if not isinstance(texto, str):
        return ""

    # Passo 1 - unicodedata.normalize("NFD", texto):
    # Separa cada caractere acentuado em duas partes: a letra base + o sinal gráfico solto.
    # Exemplo: a letra "ã" é dividida em "a" + "~"; a letra "é" vira "e" + "´".
    #
    # Passo 2 - for c in ... if unicodedata.category(c) != "Mn":
    # Percorre cada pedaço 'c' dessa string decomposta e verifica sua categoria Unicode.
    # "Mn" significa "Mark, Nonspacing" (marcas que não ocupam espaço próprio, ou seja, acentos, cedilha, til).
    # O filtro mantém apenas o que NÃO for acento (as letras puras).
    #
    # Passo 3 - "".join(...):
    # Junta todas as letras puras que sobraram de volta em uma única string limpa (ex: "Sao Paulo").
    sem_acento = "".join(
        c for c in unicodedata.normalize("NFD", texto)
        if unicodedata.category(c) != "Mn"
    )

    # Remove eventuais espaços em branco no início e no fim (.strip())
    # e transforma todas as letras em minúsculas (.lower()), retornando "sao paulo".
    return sem_acento.strip().lower()


# --- Aplicando no filtro ---

# Simula o texto digitado pelo usuário (sem acento e tudo em minúsculo).
entrada_usuario = "sao paulo"

# Passa a entrada do usuário pela nossa função de limpeza para garantir o mesmo padrão da tabela.
busca_limpa = normalizar_texto(entrada_usuario)

# 1. df["cidade"].apply(normalizar_texto): O método .apply() pega cada célula da coluna "cidade"
#    do DataFrame e a envia individualmente para a função normalizar_texto().
#    Assim, "São Paulo" na tabela se transforma temporariamente em "sao paulo" na memória.
# 2. == busca_limpa: Compara a coluna já normalizada com a busca limpa do usuário,
#    gerando uma máscara booleana (uma série de True e False).
filtro = df["cidade"].apply(normalizar_texto) == busca_limpa

# Usa a máscara booleana dentro de df[...] para exibir apenas as linhas completas
# onde a cidade correspondeu exatamente à busca (resultado True).
print(df[filtro])
```

Bloco 6: loc e iloc

```python
# .iloc (Integer Location): seleciona dados estritamente pela POSIÇÃO NUMÉRICA (índices inteiros de 0 a N-1),
# ignorando os nomes das colunas ou rótulos do índice.
# Aqui, pega a primeira linha inteira da tabela (posição 0) e a exibe como uma Series vertical.
print(df.iloc[0])             # linha pela posição

print()
# Fatiamento (slicing) por posição numérica com .iloc:
# Assim como nas listas do Python e arrays do NumPy, o índice inicial (1) é INCLUSO
# e o índice final (3) é EXCLUSO. Portanto, seleciona apenas as linhas das posições 1 e 2 (a 2ª e a 3ª linha).
print(df.iloc[1:3])           # linhas 1 e 2

print()
# .loc (Label Location): seleciona dados pelos RÓTULOS (nomes) do índice e das colunas.
# O formato é sempre df.loc[rotulo_da_linha, rotulo_da_coluna].
# Aqui, busca a linha cujo índice tem o nome/rótulo 0 e extrai especificamente o valor da coluna "nome".
print(df.loc[0, "nome"])      # linha 0, coluna "nome"

print()
# Uso profissional mais comum do .loc: filtrar linhas E selecionar colunas específicas ao mesmo tempo.
# 1. Antes da vírgula (linhas): df["idade"] > 30 filtra apenas as linhas onde a idade é maior que 30.
# 2. Depois da vírgula (colunas): ["nome", "salario"] limita a saída para mostrar somente essas duas colunas,
#    descartando o restante da tabela e deixando o resultado limpo e direto.
print(df.loc[df["idade"] > 30, ["nome", "salario"]])
```

### Resumo prático para nunca confundir .iloc e .loc:
- iloc (i = integer / inteiro): Só aceita números de posição (0, 1:3, 0:2). No slicing 1:3, o último número fica de fora (padrão Python).

- loc (l = label / rótulo): Usa o nome do índice e os nomes das colunas ("nome", ["nome", "salario"]) ou filtros booleanos (df["idade"] > 30). Se você fizer um fatiamento por rótulo no .loc (ex: df.loc[1:3]), o último item é incluído.

- iloc usa posições (números).
- loc usa rótulos e aceita condições.

Bloco 7: criando e alterando colunas

```python
# Bloco 7: criando e alterando colunas
# No pandas, criar ou atualizar colunas funciona exatamente como em um dicionário Python:
# df["nome_da_coluna"] = calculo_vetorizado. Se a coluna não existir, o pandas a cria no final da tabela;
# se já existir, ele sobrescreve os valores antigos pelos novos.

# 1. Criação de coluna numérica (Feature Engineering):
# Multiplica de forma vetorizada (sem loops) todos os valores da coluna "salario" por 12
# e grava o resultado em uma nova coluna chamada "salario_anual".
df["salario_anual"] = df["salario"] * 12

# 2. Criação de coluna booleana (bandeira/flag lógica):
# Avalia a condição (>= 35) para cada linha da coluna "idade" de uma só vez.
# Cria uma nova coluna "adulto_maduro" preenchida apenas com True (se idade for 35 ou mais) ou False (se for menor).
df["adulto_maduro"] = df["idade"] >= 35

# 3. Alteração (sobrescrita) de coluna existente:
# Como a coluna "salario" já existe no DataFrame, o pandas não cria uma nova — ele atualiza a própria coluna.
# Pega o salário atual de cada linha, multiplica por 1.10 (aplicando um reajuste de 10%) e salva por cima.
# Atenção: como essa linha vem DEPOIS do cálculo de "salario_anual", o salário anual calculado acima
# ainda reflete o valor antigo (antes do aumento de 10%). A ordem das linhas importa!
df["salario"] = df["salario"] * 1.10   # aumento de 10%

# Exibe o DataFrame completo já com as duas novas colunas adicionadas e os salários atualizados.
print(df)
```

Sem loop: a conta é feita na coluna inteira.


Bloco 8: ordenando

```python
# Bloco 8: ordenando
# O método .sort_values() reorganiza as linhas do DataFrame com base nos valores de uma ou mais colunas.
# Detalhe fundamental: por padrão, .sort_values() NÃO altera o DataFrame original (df) na memória;
# ele devolve uma cópia ordenada apenas para exibição ou para ser salva em outra variável.

# 1. Ordenação simples por uma única coluna:
# Ordena a tabela inteira olhando para a coluna "salario".
# O parâmetro ascending=False define a ordem DECRESCENTE (do maior salário para o menor).
# Se ascending=True fosse usado (ou omitido, já que True é o padrão), ordenaria do menor para o maior.
print("Ordenação pela coluna Salário")
print("="*29)
print(df.sort_values("salario", ascending=False))

# 2. Ordenação composta por múltiplas colunas (critério de desempate):
# Passando uma lista de colunas ["cidade", "idade"], o pandas ordena por hierarquia:
# - Primeiro critério: agrupa e ordena todas as linhas em ordem alfabética pela "cidade" (A -> Z).
# - Segundo critério (desempate): dentro de cada cidade repetida (ex: todas as pessoas de "São Paulo"),
#   ordena essas pessoas pela "idade" em ordem crescente (do mais jovem para o mais velho).
print()
print("Ordenação por duas colunas: Cidade e Idade")
print("="*42)
print(df.sort_values(["cidade", "idade"]))
```

Bloco 9: agrupando

```python
# Bloco 9: agrupando
# O método .groupby() segue o conceito clássico "Split-Apply-Combine" (Dividir-Aplicar-Combinar),
# muito parecido com a Tabela Dinâmica (Pivot Table) do Excel ou o GROUP BY do SQL:
# 1. Split (Dividir): separa a tabela em grupos internos com base nos valores únicos de uma coluna (ex: cada cidade).
# 2. Apply (Aplicar): calcula uma operação matemática separadamente dentro de cada grupo.
# 3. Combine (Combinar): junta os resultados em uma nova tabela resumida.

# 1. Agrupamento com uma única métrica:
# - df.groupby("cidade"): agrupa todas as linhas que têm a mesma cidade.
# - ["salario"]: seleciona apenas a coluna de salário dentro de cada grupo para fazer a conta.
# - .mean(): calcula a média salarial de cada cidade individualmente e exibe o resumo.
print(df.groupby("cidade")["salario"].mean())

# 2. Agrupamento com múltiplas métricas simultâneas usando .agg() (Aggregate):
# Em vez de rodar um comando para a média e outro para a contagem, o método .agg() permite passar
# uma lista de funções estatísticas para serem calculadas de uma única vez sobre a coluna "salario" de cada cidade.
# Aqui, retorna uma tabela com duas colunas: "mean" (média salarial da cidade) e "count" (quantas pessoas moram nela).
print()
print(df.groupby("cidade")["salario"].agg(["mean", "count"]))
```

O groupby responde perguntas como "qual o salário médio por cidade?". Em Excel seria uma tabela dinâmica.

### Por que o .groupby() com .agg() é ouro na Engenharia de IA:
- Quando você precisa criar novas variáveis (features) para um modelo de Machine Learning, raramente os dados brutos bastam.
- Com o .agg(), você extrai o perfil completo de cada grupo em uma linha — por exemplo, calcular ao mesmo tempo o ticket médio ("mean"), o maior valor ("max"), a dispersão ("std") e o volume total ("sum") de compras por cliente ou por categoria:

```python
df.groupby("cidade")["salario"].agg(["mean", "min", "max", "count"])
```

Bloco 10: dados faltantes

```python
# Bloco 10: dados faltantes
# Em dados reais (planilhas, bancos de dados, retornos de APIs), células vazias são representadas
# pelo pandas como NaN (Not a Number) para colunas numéricas ou None/NaN para textos.
# Saber diagnosticar e tratar esses valores é obrigatório antes de alimentar qualquer modelo de IA.

# Cria um DataFrame de exemplo com 3 linhas e 2 colunas ("x" numérica e "y" de texto),
# inserindo propositalmente valores ausentes (None) em posições diferentes para simular dados sujos.
df2 = pd.DataFrame({
    "x": [1, None, 3], # coluna 1: números
    "y": ["a", "b", None]  # coluna 2: textos
})

# 1. Diagnóstico célula por célula (.isna()):
# Percorre toda a tabela e retorna um novo DataFrame de mesmo tamanho contendo apenas True e False:
# - True: a célula está vazia/nula (NaN).
# - False: a célula possui um valor válido preenchido.
print(df2.isna())

print()
# 2. Resumo de valores faltantes por coluna (.isna().sum()):
# Como o Python trata True como 1 e False como 0, encadear .sum() logo após .isna()
# soma os valores True de cada coluna de cima para baixo (axis=0).
# Resultado: mostra exatamente quantos dados estão faltando em cada coluna (x: 1, y: 1).
print("Mostrar os dados faltantes de cada coluna")
print(df2.isna().sum())

print()
# 3. Remoção de linhas incompletas (.dropna()):
# Varre a tabela e apaga qualquer linha inteira que tenha PELO MENOS UM valor faltante (NaN).
# No exemplo, apenas a primeira linha (índice 0: x=1.0, y="a") está 100% completa, então só ela sobra.
# Importante: assim como o .sort_values(), o .dropna() não altera 'df2' original; devolve uma cópia limpa.
print("Removendo a linha inteira se existir valor faltante")
print(df2.dropna())

print()
# 4. Preenchimento (imputação) de valores faltantes (.fillna()):
# Em vez de jogar a linha inteira fora e perder dados úteis da outra coluna, o .fillna(0)
# substitui todas as células vazias (NaN) da tabela pelo valor informado entre parênteses (neste caso, 0).
# Na prática de Engenharia de Dados, costuma-se preencher por coluna específica:
# ex: df2["x"] = df2["x"].fillna(df2["x"].mean()) para números e df2["y"] = df2["y"].fillna("Desconhecido") para textos.
print("Preenchendo valores faltantes")
print(df2.fillna(0))
```

Dados reais sempre têm buracos. Saber detectá-los e tratá-los é essencial.

### Detalhe importante sobre o tipo de dado no Bloco 10:
- Ao criar a coluna "x" com [1, None, 3], o pandas converte automaticamente os números inteiros 1 e 3 para decimais (1.0 e 3.0, do tipo float64).
- Isso acontece porque, historicamente no NumPy/pandas, o marcador NaN é um valor especial de ponto flutuante (float), então a coluna inteira assume esse tipo para conseguir armazenar o dado faltante.

```python
produtos = pd.DataFrame({
    "produto": ["Camiseta", "Calça", "Tênis", "Boné", "Jaqueta", "Meia"],
    "categoria": ["Roupa", "Roupa", "Calçado", "Acessório", "Roupa", "Acessório"],
    "preco": [49.9, 120.0, 250.0, 35.0, 300.0, 15.0],
    "estoque": [30, 12, 8, 50, 5, 100]
})
```

1. Mostre o formato (shape) e as estatísticas com describe().
2. Mostre só as colunas produto e preco.
3. Filtre os produtos com preço maior ou igual a 100.
4. Crie a coluna valor_estoque (preço × estoque) e mostre o produto com maior valor em estoque (pesquise idxmax).
5. Mostre o preço médio por categoria.
6. Desafio: mostre os produtos da categoria Roupa com estoque menor que 20, ordenados por preço do maior para o menor.

Se travar: leia o erro, tente por 10 minutos e só então pesquise. Erro comum: usar and em vez de &, ou esquecer os colchetes duplos ao selecionar várias colunas.

```python
# 1. Mostre o formato (shape) e as estatísticas com describe().
print(produtos.shape)
print()
print(produtos.describe())
```

```python
# 2. Mostre só as colunas produto e preco.
print(produtos[["produto", "preco"]])
```

```python
# 3. Filtre os produtos com preço maior ou igual a 100.
print(produtos[produtos['preco'] >= 100])
```

```python
# 4. Crie a coluna valor_estoque (preço × estoque) e mostre o produto com maior valor em estoque (pesquise idxmax).

# 1. Criação vetorizada da coluna (correto)
produtos["valor_estoque"] = produtos["preco"] * produtos["estoque"]

# 2. Captura o índice (posição/rótulo da linha) do maior valor em estoque
indice_maior = produtos["valor_estoque"].idxmax()

# 3. Usa o .loc[] para buscar os dados do produto naquela linha específica:
print(produtos.loc[indice_maior])            # Mostra a linha inteira (nome, categoria, preco, estoque, valor_estoque)
print()
print(produtos.loc[indice_maior, "produto"])    # Mostra apenas o nome do produto
```

```python
# 1. Agrupa as linhas pela coluna "categoria"
# 2. Seleciona a coluna "preco" dentro de cada grupo
# 3. Calcula a média (.mean()) para cada categoria individualmente
preco_medio_categoria = produtos.groupby("categoria")["preco"].mean()

# Formata cada valor da Series como texto monetário com 2 casas decimais
print(preco_medio_categoria.map("R$ {:.2f}".format))
```

```python
# 6. Desafio: mostre os produtos da categoria Roupa com estoque menor que 20, ordenados por preço do maior para o menor.

# 1. Filtra: categoria igual a "Roupa" E estoque menor que 20
filtro = (produtos["categoria"] == "Roupa") & (produtos["estoque"] < 20)
roupas_baixo_estoque = produtos[filtro]

# 2. Ordena o resultado pelo preço em ordem decrescente (do maior para o menor)
resultado = roupas_baixo_estoque.sort_values("preco", ascending=False)

print(resultado)
```

### Como escrever tudo em uma única expressão (estilo pipeline):
Quando você já estiver confortável com as etapas separadas, também pode encadear os métodos diretamente usando .loc[]:

```python
resultado = (
    produtos
    .loc[(produtos["categoria"] == "Roupa") & (produtos["estoque"] < 20)]
    .sort_values("preco", ascending=False)
)

print(resultado)
```
